#### Merge Function 1 Example


```python
import pandas as pd

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1)

df1 = pd.DataFrame(data2)


print(df ,"\n""\n", df1)
```

```python
# Merging Data Frame using one Key

result = pd.merge(df , df1 , on = "key")
result
```

```python
"""
MERGE 	    Description
left		Use keys from left frame only
right		Use keys from right frame only
outer		Use union of keys from both frames
inner       Use intersection of keys from both frames
"""
```

```python
import pandas as pd

data1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K1', 'K0', 'K1'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

data2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K0', 'K0', 'K0'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(data1)

df1 = pd.DataFrame(data2)


print(df, "\n\n", df1)
```

```python
# Merge LEFT with multiple Key
pd.merge(df, df1, how = "left", on = ["key","key1"])
```

```python
# Merge RIGHT with multiple Key
pd.merge(df,df1, how="right", on=["key","key1"])
```

```python
# Merge OUTER with multiple Key
pd.merge(df,df1, how = "outer", on = ["key","key1"])
```

```python
# Merge INNER with multiple Key
pd.merge(df,df1, how = "inner", on = ["key","key1"])
```

Lambda Function 2 examples with Map, Filter, Reduce


# map() 
The map() function applies a function to every item in an iterable:

```python
# 1. Find the length of each word

words = ["python", "data", "science", "pandas"]

length = list(map(lambda x : len(x), words))
length
```

```python
# 2 Convert all temperatures from Celsius to Fahrenheit.

celsius = [0, 10, 20, 30, 40]

# Formula = (C * 9/5) + 32

fahrenheit = list(map(lambda x : ((x*9/5)+32), celsius))
fahrenheit
```

# Filter()
The filter() function creates a list of items for which a function returns True:

```python
#1.Extract all numbers greater than 50.

numbers = [10, 55, 23, 78, 90, 45, 67]

num = list(filter(lambda x : x > 50 , numbers))
num
```

```python
# 2.Keep only words that start with the letter 'S'.

names = ["Smriti", "Aradhya", "Swara", "Dinesh", "Sanjana"]

updated_names = list(filter(lambda x : x[0]=="S", names))
updated_names
```

# reduce()
Returns the output with one single value.

```python
from functools import reduce

# 1.Find the product of all numbers.
numbers = [2, 3, 4, 5]

result = reduce(lambda x,y : x*y, numbers)
result
```

```python
# 2.Find the largest number using reduce().

numbers = [25, 8, 90, 12, 45]

large_num = reduce(lambda x,y : x if x>y else y, numbers)
large_num
```
