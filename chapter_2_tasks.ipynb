# SETS

```python
data = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

print(type(data))
print(dir(data))
```

```python
# add

data = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

data.add("science")
data
```

```python
# clear

data.clear()
data
```

```python
# copy

data = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

data_1 = data.copy()
data_1
```

```python
# difference

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}

data_3 = data_1.difference(data_2)
print(data_3)
```

```python
# difference_update

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}  
data_2 = {11, 20, "AI", "Machine"}

data_1.difference_update(data_2) # remove the difference and update the original set
data_1
```

```python
# discard

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

data_1.discard("artificial")
data_1.discard("Python")
data_1
```

```python
# intersection

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

data_2 = {11, 20, "AI", "Machine"}

data_3 = data_1.intersection(data_2) #Return the intersection of two sets as a new set
data_3
```

```python
#intersection_update


data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

data_2 = {11, 20, "AI", "Machine"}

data_1.intersection_update(data_2) #Update a set with the intersection of itself and another
data_1
```

```python
# isdisjoint

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}
data_3 = {1,2,3}
print(data_1.isdisjoint(data_2)) # Return True if two sets have a null intersection
print(data_1.isdisjoint(data_3))
```

```python
# issubset

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}

print(data_1.issubset(data_2)) # checks whether all elements of one set are present in another set.

a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))
```

```python
# issuperset

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}
print(data_1.issuperset(data_2))

a = {1, 2}
b = {1, 2, 3, 4}

print(b.issuperset(a))
```

```python
# pop

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

print(data_1.pop())
print(data_1)
```

```python
# remove

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}

data_1.remove(10)
print(data_1)
```

```python
# symmetric_difference

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}  # Return the symmetric difference of two sets as a new set.
data_2 = {11, 20, "AI", "Machine"}

data_3 = data_1.symmetric_difference(data_2)

data_3
```

```python
# symmetric_difference_update

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}

data_1.symmetric_difference_update(data_2) # Update a set with the symmetric difference of itself and another
print(data_1)
```

```python
# union

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}

data_3 = data_1.union(data_3) # Return the union of sets as a new set.
data_3
```

```python
# update

data_1 = {10, 20, 30, "Python", "AI", 10.5, (1, 2)}
data_2 = {11, 20, "AI", "Machine"}

data_1.update(data_2) #Update a set with the union of itself and others.
print(data_1)
```

```python
# clear()

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

employee.clear()

employee
```

```python
# copy

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

employee_1 = employee.copy()

employee_1

```

```python
# fromkeys

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

new_employee = employee.fromkeys(employee,0) # Create a new dictionary with keys from iterable and values set to value
new_employee
```

```python
# get

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

print(employee.get("age")) # Return the value for key if key is in the dictionary, else default.

print(employee.get("time")) # Return the value for key if key is in the dictionary, else default.
```

```python
# items

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

employee.items() # a set-like object providing a view on items
```

```python
# keys
employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}
employee.keys()
```

```python
# pop
employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

print(employee.pop("age")) # age is removed , Remove a specific key from a dictionary
print(employee)
#print(employee.pop("time")) # if the key is not found , raise a key error.
```

```python
# popitem

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

print(employee.popitem()) # Remove and return a (key, value) pair as a 2-tuple.
print(employee)
```

```python
# setdefault --> Insert key with a value of default if key is not in the dictionary.Return the value for key if key is in the dictionary, else default

employee = {
    "emp_id":301221,
    "email":"dinesh@gmail.com",
    "company":"Reliance",
    "Location":"Bangalore",
    "experience":7,
    "age":30,
    "married":True
}

employee.setdefault("department","Telecom")
employee.setdefault("salary")
employee.setdefault("age")
print(employee)
```

### Create 3 Data frames using dictionaries.

```python
import pandas as pd

employee_data = {
    "names":["Dinesh","Manju","Tamil","Nanda"],
    "age":[30,29,27,30],
    "emp_id":[301,302,303,304],
    "company":["Radisys","SBI","TVS","DLF"]
} 

df = pd.DataFrame(employee_data)
print(df)
```

```python
# created Data frame using dict, but with custom index

import pandas as pd

employee_data = {
    "names":["Dinesh","Manju","Tamil","Nanda"],
    "age":[30,29,27,30],
    "emp_id":[301,302,303,304],
    "company":["Radisys","SBI","TVS","DLF"]
} 

df = pd.DataFrame(employee_data,index = ["emp_1","emp_2","emp_3","emp_4"])

print(df)
```
