```python
# Calculate the percentage of a student.

class Student():
    institute = "GUVI"
    def __init__(self,username,userroll,course,marks):
        self.username = input("Enter your name: ")
        self.userroll = int(input("Enter your roll no: "))
        self.course = input("Enter your course: ")
        self.marks = int(input("Enter your marks: "))

    def percentage(self,percentage):
        max_mark = 140
        self.percentage = ((self.marks)/max_mark*100)

s1 = Student("username","userroll","course","marks")
s2 = Student("username","userroll","course","marks")
s1.percentage("percentage")
s2.percentage("percentage")

print(f"institute is {s1.institute}, NAME is {s1.username}, roll is {s1.userroll}, course is {s1.course}, mark is {s1.marks}, percentage is {s1.percentage}")
print(f"institute is {s2.institute}, NAME is {s2.username}, roll is {s2.userroll}, course is {s2.course}, mark is {s2.marks}, percentage is {s1.percentage}")




```

```python
class Student():
    institute = "GUVI"
    def __init__(self):
        self.username = input("Enter your name: ")
        self.userroll = int(input("Enter your roll no: "))
        self.course = input("Enter your course: ")
        self.marks = int(input("Enter your marks: "))

    def percentage(self):
        max_mark = 140
        return round((self.marks)/max_mark*100,2)

s1 = Student()
s2 = Student()


print(f"institute is {s1.institute}, NAME is {s1.username}, roll is {s1.userroll}, course is {s1.course}, mark is {s1.marks}, percentage is {s1.percentage()}")
print(f"institute is {s2.institute}, NAME is {s2.username}, roll is {s2.userroll}, course is {s2.course}, mark is {s2.marks}, percentage is {s1.percentage()}")
```

```python
"""
Python class named car that models the car with below specifications.

Attributes:
make(string) : Represent the manufacturer of the car (eg : toyota, honda)
model(string): Represent the model of the car (eg: xuv 500, xuv 700)
year(integer): Represent the year of the manufacturer.

Methods/functions:
start() --> prints the car is starting
stop() --> prints the car is stopping.
get_info() --> Returns the string containing the car's make, model, and year in the format  : "Make is {make}, Model is {model}, year is {year}"

"""

class Car:
    def __init__(self):
        self.make = input("Enter your car manufacturer: ")
        self.model = input("Enter your car model: ")
        self.year = int(input("Enter your car manufactured year: "))

    def start(self):
        print("The car is starting")

    def stop(self):
        print("The car is stopping")

    def get_info(self):
        return f"Make is {self.make}, Model is {self.model}, Year is {self.year}"


car1 = Car()
car2 = Car()

print(car1.get_info())
print(car2.get_info())
```

### # Concept --> Banking System

```python

# Abstraction
"""
In the below code, I have created Abstraction class called BankAccount, that defines what all the bank account must do.
"""

from abc import ABC,abstractmethod

class BankAccount(ABC):
    @abstractmethod
    def cash_withdrawal(self):
        pass

    @abstractmethod
    def account_type(self):
        pass

#Inheritance
# now im creating a child class and inherit the parant class BankAccount.

class SBI_bank(BankAccount):
    def cash_withdrawal(self,amount):
        return f"The amount withdrawn in saving acc is {amount}"

    def account_type(self):
        return "The account type is savings"

class HDFC_bank(BankAccount):
    def cash_withdrawal(self,amount):
        return f"The amount withdrawn in current acc is {amount}"

    def account_type(self):
        return f"The account type is current"

#Polymorphism
# I have created a method "cash_withdrawal" same method but with different operation in different banks.

acc1 = SBI_bank()
acc2 = HDFC_bank()

print(acc1.cash_withdrawal(1000))
print(acc1.account_type())

print(acc2.account_type())
print(acc2.cash_withdrawal(1500))





```
