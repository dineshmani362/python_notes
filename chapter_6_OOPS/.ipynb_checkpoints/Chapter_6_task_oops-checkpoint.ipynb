### Concept --> Banking System

```python
# Abstraction
"""
In the below code, I have created Abstraction class called BankAccount, that defines what all the bank account must do.
"""

from abc import ABC,abstractmethod

class BankAccount(ABC):
    @abstractmethod
    def cash_withdrawal(self):
        pass

    @abstractmethod
    def account_type(self):
        pass

#Inheritance
# now im creating a child class and inherit the parant class BankAccount.

class SBI_bank(BankAccount):
    def cash_withdrawal(self,amount):
        return f"The amount withdrawn in saving acc is {amount}"

    def account_type(self):
        return "The account type is savings"

class HDFC_bank(BankAccount):
    def cash_withdrawal(self,amount):
        return f"The amount withdrawn in current acc is {amount}"

    def account_type(self):
        return f"The account type is current"

#Polymorphism
# I have created a method "cash_withdrawal" same method but with different operation in different banks.

acc1 = SBI_bank()
acc2 = HDFC_bank()

print(acc1.cash_withdrawal(1000))
print(acc1.account_type())

print(acc2.account_type())
print(acc2.cash_withdrawal(1500))
```
