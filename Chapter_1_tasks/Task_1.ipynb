# Task 1

## Give 2 examples of all String functions which should be meaningful.

```python
first_name = "dineshmani"
last_name = "SUNDARAM"
dir(first_name)
```

```python
# String Operation = capitalize
first_name = "dineshmani"
first_name.capitalize()
```

```python
# String Operation = casefold
last_name = "SUNDARAM"
last_name.casefold()
```

```python
# String Operation = center
first_name = "dineshmani"
first_name.center(20,'*')
```

```python
# String Operation = count
first_name = "dineshmani"
first_name.count('i')
```

```python
# String Operation = encode
first_name = "dineshmani"
data = first_name.encode()
print(data)
print(type(data))
```

```python
# String Operation = endswith
first_name = "dineshmani"

print(first_name.endswith('i'))
print(last_name.endswith('d'))
```

```python
# String Operation = expandtabs
name = "Dinesh\tmani"
print(name.expandtabs(4))
print(name.expandtabs(6))
```

```python
# String Operation = find
Name = "Dineshmani"
print(Name.find('D'))
print(Name.find('i'))
```

```python
# String Operation = format
Name = "My first name is Dinesh and last name is {}"
Name.format('mani')
```

```python
# string operation = format_map
Name = "My first name is {fname} and last name is {lname}"
data = {
    'fname':'Dinesh',
    'lname':'mani'
}
Name.format_map(data)
```

```python
# String Operation = index

client_name = "Samsung"
client_name.index('m')
```

```python
# string operation = isalnum (checks whether all characters in the string are:alphabets (A-Z, a-z) or numbers (0-9))

client_name = "Reliance"
client_name_1 = "Reliance@123"
print(client_name.isalnum())
print(client_name_1.isalnum())
```

```python
# string operation = isalpha, Return True if the string is an alphabetic string, False otherwise.
Name = "Subramani_1"
Name_1 = "Dinesh"
print(Name.isalpha())
print(Name_1.isalpha())
```

```python
# string operation = isascii, Return True if all characters in the string are ASCII, False otherwise.
Name = "Dineshmani"
empty_string = " "
text = "தமிழ்"
print(empty_string.isascii())
print(Name.isascii())
print(text.isascii())
```

```python
# string operation = isdecimal, Return True if the string is a decimal string, False otherwise.
value = "12345"
value_1 = "2.0"
print(value.isdecimal())
print(value_1.isdecimal())
```

```python
# string operation = isdigit, Return True if the string is a digit string, False otherwise.
value = "12345"
value_1 = "2.0"
print(value.isdigit())
print(value_1.isdigit())
```

```python
# string operation = isidentifier , it should not start with number, it should not contain space.
value = "12345"
value_1 = "Dinesh_1"
value_2 = "Dinesh mani" 
print(value.isidentifier())
print(value_1.isidentifier())
print(value_2.isidentifier())
```

```python
# string operation = islower
name = "DINESH"
name_1 = "dinesh"
print(name.islower())
print(name_1.islower())
```

```python
# string operarion = isnumeric, Return True if the string is a numeric string, False otherwise.

password = "Dinesh123"
password_1 = "12344"
print(password.isnumeric())
print(password_1.isnumeric())
```

```python
# string operation = isprintable, Return True if the string is printable, False otherwise.
course = "Artificial Intelligence"
greeting = "Hello\nworld"
print(course.isprintable())
print(greeting.isprintable())
```

```python
# string operation = isspace, Return True if the string is a whitespace string, False otherwise.
course = "Artificial Intelligence"
empty_string = "    "
print(course.isspace())
print(empty_string.isspace())
```

```python
# string operation = istitle, Return True if the string is a title-cased string, False otherwise.
course = "Artificial Intelligence"
course_1 = "machine learning"
print(course.istitle())
print(course_1.istitle())
```

```python
# string operation = isupper, return True if the string is an uppercase string, False otherwise.
course = "Artificial Intelligence"
course_1 = "MACHINE LEARNING"
print(course.isupper())
print(course_1.isupper())
```

```python
# string operartion = join
Name = "Dineshmani"
print("-".join(Name))
```

```python
# string operatrion = ljust, Return a left-justified string of length width.
Name = "Dinesh"
print(Name.ljust(10))
print(Name.ljust(10,'*'))
```

```python
# string operation = lower, Return a copy of the string converted to lowercase
course = "PYTHON"
course.lower()
```

```python
# string operation = lstrip, Return a copy of the string with leading whitespace removed.
course = "PYTHON     "
course_1 = "    AI"
print(course.lstrip())
print(course_1.lstrip())
```

```python
# string operarion = maketrans(),translate(), maketrans() only creates the mapping table.To apply the replacement, use translate()

course = "Artificial Intelligence"
course_1 = "Machine Learning"
table = str.maketrans('ail','@IL')
print(course.translate(table))
print(course_1.translate(table))
```

```python
# string operation = partition, Partition the string into three parts using the given separator.
course = "Artificial Intelligence"
fruits = "apple-mango-orange"
print(fruits.partition('-'))
print(course.partition(' '))
```

```python
# string operation = removeprefix, Return a str with the given prefix string removed if present.
Name = "Mr.Dinesh"
Name.removeprefix('Mr.')
```

```python
# string operation = removesuffix, Return a str with the given suffix string removed if present.
Name = "Mr.Dineshmani"
Name.removesuffix("mani")
```

```python
# string operation = replace, Return a copy with all occurrences of substring old replaced by new.
course = "Artificial Intelligence"
course.replace("Artificial Intelligence","Machine Learning")
```

```python
# string operation = rfind, Return the highest index, return -1 if not present.

course = "Artificial Intelligence"
print(course.rfind('e'))
print(course.rfind('A'))
print(course.rfind('a'))
print(course.rfind('A',2,5))
print(course.rfind('l',2,22))
```

```python
# string operation = rindex
course = "Artificial Intelligence"
print(course.rindex('e'))
print(course.rindex('A'))
print(course.rindex('a'))
print(course.rindex('l',2,22))
```

```python
# string operation = rjust
course = "Artificial Intelligence"
course.rjust(30)
```

```python
# string operation = rpartition, Partition the string into three parts using the given separator.
course = "Artificial - Intelligence"
course.rpartition('-')
```

```python
# string operation = rsplit, Return a list of the substrings in the string, using sep as the separator string
fruits = "apple-banana-orange"
print(fruits.rsplit('-'))
print(fruits.rsplit('-',1))
```

```python
# string operation = rstrip, Return a copy of the string with trailing whitespace removed.
fruits = "apple     "
fruits.rstrip()
```

```python
# string operation = split, Return a list of the substrings in the string, using sep as the separator string.
fruits = "apple-banana-orange"
print(fruits.split('-'))
print(fruits.split('-',1))
```

```python
# string operation = splitlines, Return a list of the lines in the string, breaking at line boundaries.
fruits = "apple\nbanana\norange"
print(fruits.splitlines())
```

```python
# string operation = startswith, Return True if String starts with the specified prefix, False otherwise
fruits = "apple-banana-orange"
print(fruits.startswith('a'))
print(fruits.startswith('b'))
```

```python
# string operation = strip, Return a copy of the string with leading and trailing whitespace removed.
course = "     AI     "
course.strip()
```

```python
# string operation = swapcase, Convert uppercase characters to lowercase and lowercase characters to uppercase
fruits = "apple-BANANA-orange"
fruits.swapcase()
```

```python
# string operation = title
fruits = "apple-BANANA-orange"
course = 'aI'
print(fruits.title())
print(course.title())
```

```python
# string operation = upper
course = "ai"
name = "dinesh"
print(course.upper())
print(name.upper())
```

```python
# string operation = zfill
name = "Dinesh"
print(name.zfill(10))
print("-25".zfill(5))
print("25".zfill(5))
```
