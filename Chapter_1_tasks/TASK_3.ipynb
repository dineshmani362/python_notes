Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.

Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using trypt or con

###dition checks.


```python
import random
while True:
    try:
        user_input = input("Enter your choice, rock or paper or scissor").lower()
        print(f"The user input is : {user_input}")
        computer = ["rock","paper","scissor"]
        if user_input not in computer:
            raise ValueError("Input is not proper, please enter proper input")
        random_select = random.choice(computer)
        print(f"The computer selected : {random_select}")

        if user_input == random_select:
            print("Match draw")
        elif user_input == "rock" and random_select == "scissor":
            print("user won the game")
        elif user_input == "paper" and random_select == "rock":
            print("user won the game")
        elif user_input == "scissor" and random_select == "paper":
            print("user won the game")
        else:
            print("Computer won the game")
    except ValueError as e:
        print(e)
    except Exception as e:
        print("something went wrong", e)
    
    play_again = input("Do you want to play again: (yes/no): ").lower()
    if play_again != "yes":
        print("Game exited")
        break
        
```
