# Word Guessing Game (Mini Hangman)
Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:
●	Loops
●	Conditional logic
●	String manipulation
●	Exception handling
●	Logical thinking

Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


```python
import random

words = ["python","machine","learning","bridge"]

selected = random.choice(words)

display = ["_"] * len(selected)

max_attempt = 6

print("Welcome to the word guessing game")

while max_attempt > 0 and "_" in display:

    print(" ".join(display))
    print("Chances left: ", max_attempt)

    letter = input("Enter the letter : ").lower()

    if letter in selected:
        
        for i in range(len(selected)):
            if selected[i] == letter:
                display[i] = letter
    else:
        max_attempt = max_attempt - 1
        print("Its wrong")

if "_" not in display:
    print("You guessed the word:", selected)
    print("You won")
else:
    print("No chances left")
    print("The word was:", selected)
```
